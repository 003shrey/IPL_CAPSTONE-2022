##Loading the Libraries and Dataset


```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")
url="https://raw.githubusercontent.com/AkarshVyas/Data-Visualization-Youtube/refs/heads/main/IPL.csv"
df = pd.read_csv( url)

df.head()
```

```python
df.info()
```

```python
#Check the size of rows and columns of the dataset
df.shape
```

```python
print(f"your shape is {df.shape[0]} and your columns are {df.shape[1]}")
```

```python
#Now let's see how many columns have null values in total.
df.isnull().sum()
```

Which team won the most matches?

```python
match_wins=df['match_winner'].value_counts()
```

```python
sns.barplot(y=match_wins.index, x=match_wins.values)
plt.title("most match win")
```

Toss Decision Trends

```python
sns.countplot(x= df["toss_decision"])
```

Toss Winner vs Match Winner

```python
count=df[df['toss_winner'] == df['match_winner']]["match_id"].count()
percentage = (count*100)/ df.shape[0]
percentage.round(2)
```

How do teams win? (Runs vs Wickets)

```python
sns.countplot(x=df["won_by"])
```

Key Player Performances

```python
count= df['player_of_the_match'].value_counts().head(10)
count
```

```python
sns.barplot(y=count.index, x=count.values, palette='mako')
plt.title("key player")
```

2 Top Scorers

```python
high =  df.groupby('top_scorer')["highscore"].sum().sort_values(ascending=False).head(2)
high
```

```python
high.plot(kind='barh')
```

10 Best Bowling Figures

```python
df['highest_wickets']= df['best_bowling_figure'].apply(lambda x :x.split('--')[0])
df['highest_wickets'] = df['highest_wickets'].astype('int')
df['highest_wickets']
```

```python
top_bowler = df.groupby('best_bowling')['highest_wickets'].sum().sort_values(ascending=False).head(10)
top_bowler
```

```python
top_bowler.plot(kind='barh')
```
